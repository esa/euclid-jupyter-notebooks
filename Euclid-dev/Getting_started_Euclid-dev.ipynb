# Getting Started with Euclid-Dev

## Development and testing environment

**Euclid-Dev** is the development and testing version of the standard **Euclid** App in the Euclid Data Space.

It uses the **same environments as the Euclid app**, including the JupyterLab interface, scientific software environments and access to public Euclid data, but also includes functionality that is being evaluated before being introduced into the stable application.

Euclid-Dev provides early access to:

- new Euclid Data Space functionality;
- new or updated scientific software;
- changes to the Euclid JupyterLab environment;
- experimental tools and extensions;
- functionality being considered for future Euclid app releases.

Because Euclid-Dev is intended for development and testing, it may be updated frequently. Functionality may occasionally be unstable, incomplete or change without notice.

> **For stable, reproducible or production workflows, use the standard Euclid application.**

Users are encouraged to test new functionality in Euclid-Dev and provide feedback through the **Euclid Helpdesk**.

## Euclid Explorer

One of the functionalities currently available for testing in Euclid-Dev is **Euclid Explorer**.

Euclid Explorer is a JupyterLab extension for interactively exploring Euclid data and producing cutouts of Euclid objects. It provides a visual way of browsing observations and available data products directly from JupyterLab before moving to more detailed analysis workflows.

Euclid Explorer can be used to:

- interactively explore Euclid observations;
- identify sources and regions of interest;
- inspect available Euclid data products;
- generate cutouts of Euclid objects;
- move from visual exploration to notebook-based scientific analysis.

As Euclid Explorer is currently provided as test functionality in Euclid-Dev, its interface and functionality may evolve.

### Opening Euclid Explorer

After launching Euclid-Dev, open **Euclid Explorer** by clicking the **Euclid Explorer icon in the left-hand sidebar of JupyterLab**. The Explorer opens as a panel alongside your notebooks, allowing you to explore Euclid products while continuing to work in the JupyterLab environment.

## Access to Euclid data

Euclid-Dev provides direct access to the **public Euclid data volumes**, using the same data access approach as the standard Euclid app.

**Euclid data volumes** are mounted when the App starts. Read products directly from these volumes rather than downloading duplicate copies into your workspace.

| Dataset | Mounted path | Size | Data product description |
|---|---|---:|---|
| **DR1-Foundation** | `/data/euclid_idr1/` | 1.6 PB | [DR1 Data Product Description Document](https://euclid.esac.esa.int/dr/dr1/dpdd/) |
| **DR1 auxiliary calibration data** | `/data/euclid_dr1_cal/` | — | [DR1 Auxiliary Data Description](https://www.cosmos.esa.int/web/euclid/dr1-auxiliary-data) |
| **Q2** | `/data/euclid_q2/` | 3.1 TB | [Q2 Data Product Description](https://www.cosmos.esa.int/web/euclid/q2-documentation) |
| **Q1** | `/data/euclid_q1/` | 35 TB | [Q1 Data Product Description Document](https://euclid.esac.esa.int/dr/q1/dpdd/) |
| **Early Release Observations (ERO)** | `/data/euc_ero_data_01/` | 600 GB | [ERO Data Product Description](https://www.cosmos.esa.int/web/euclid/ero-public-release) |

### Euclid Astroquery

Use the pre-installed [Euclid Astroquery package](https://astroquery.readthedocs.io/en/latest/esa/euclid/euclid.html) to query the Archive and identify relevant data products. The examples demonstrate cone searches, catalogue queries and product access.

Use the `datalabs_path` values returned by the relevant Archive queries to locate products on the mounted data volumes. Consult each notebook for the query and product-specific workflow.

## Example Jupyter notebooks

A collection of example notebooks is provided to help you get started with Euclid data analysis. The notebooks are maintained in two public GitHub repositories:

### ESA-developed notebooks

[esa/euclid-jupyter-notebooks](https://github.com/esa/euclid-jupyter-notebooks)

These notebooks provide examples and tutorials developed for working with the Euclid Data Releases.

### Community-contributed notebooks

[esa/euclid-contributed-jupyter-notebooks](https://github.com/esa/euclid-contributed-jupyter-notebooks)

These notebooks are contributed by members of the scientific community and curated by the Euclid Data Space user Group. They and provide advanced workflows, science examples and tools for working with Euclid data.

> **Community contribution notice:** Community-contributed notebooks are provided by their respective developers and are not maintained by ESA. They may require additional packages or updates and may not always work out of the box.
