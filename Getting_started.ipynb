This file contains some helpful information to get you started on accessing and analysing data from the **ESA Euclid Mission** on **Euclid Data Space**.

An **App** is a purpose-built software application available for Data Analysis. From the user's perspective, it functions similarly to a virtual machine, allowing users to launch, use, and delete it as needed. For more information on the functionalities of the datalab implementation, please read [here](https://datalabs.esa.int/help/Datalabs.html). The **Euclid App** is based on JupyterLab. For basics on JupyterLab, you can access here the [JupyterLab documentation](https://jupyterlab.readthedocs.io/en/stable/).

The **Euclid App** provides direct access to **Euclid DR1-Foundation data** on Euclid Data Space, allowing many analyses to be performed without first downloading the data. Earlier Euclid public releases, including **Q1, Q2 and the Early Release Observations (EROs)**, are also available. 

> **Analyse data where they are stored.** Use the mounted Euclid data volumes for analysis. To discover and download Euclid data to your own computer, use the [Data Discovery tab](https://euclid.dataspace.esa.int/data-discovery/basic-search).

## 1. Workspaces

### Storage and access

Data Analysis provides access to `my_workspace` and `team_workspaces`:

- **My Workspace:** Your personal folder, with 100 GB of storage.
- **Team Workspaces:** Shared spaces for collaboration and analysis requiring more storage. To request a team workspace, contact the [Euclid Helpdesk](https://support.cosmos.esa.int/euclid/). Changes made in a team workspace are visible to other users with access to that workspace.

> **Save your work in a workspace.** Files stored only in the App's local filesystem are deleted when the App instance is deleted. Copy notebooks, environment files and analysis outputs to `my_workspace` or an appropriate team workspace.

### Uploading and downloading your files

- **JupyterLab:** Use the upload and download controls for individual files. Multiple files must be zipped before downloading them together.
- **Workspaces menu:** Use the folder icon at the top right for workspace transfers, including the option to zip files before downloading.
- **FileZilla:** Use the FileZilla App as an alternative transfer method. See the [workspace transfer guide](https://datalabs.esa.int/s2e2-help/Workspaces.html#heading_toc_j_5) for illustrated instructions.

### Resources and monitoring

- Up to **two Apps** running simultaneously.
- Default resources for the Euclid App: **32 GB RAM** and **2 CPU cores**. Can be customised using the **Advanced** configuration when launching the App. 
- User quota: **64 GB RAM** and **16 CPU cores**.
- **100 GB** of personal workspace storage.

For general resource usage, open **My resource limits**. The JupyterLab status bar at the bottom shows memory usage across kernels and terminals. For additional diagnostics, run `diagnose` in a terminal.


## 3. How to access Euclid data

### Mounted Euclid data volumes

**Euclid data volumes** are mounted when the App starts. Read products directly from these volumes rather than downloading duplicate copies into your workspace.

| Dataset | Mounted path | Size | Data product description |
|---|---|---:|---|
| **DR1-Foundation** | `/data/euclid_idr1/` | 1.6 PB | [DR1 Data Product Description Document](https://euclid.esac.esa.int/dr/dr1/dpdd/) |
| **DR1 auxiliary calibration data** | `/data/euclid_dr1_cal/` | — | [DR1 Auxiliary Data Description](https://www.cosmos.esa.int/web/euclid/dr1-auxiliary-data) |
| **Q2** | `/data/euclid_q2/` | 3.1 TB | [Q2 Data Product Description](https://www.cosmos.esa.int/web/euclid/q2-documentation) |
| **Q1** | `/data/euclid_q1/` | 35 TB | [Q1 Data Product Description Document](https://euclid.esac.esa.int/dr/q1/dpdd/) |
| **Early Release Observations (ERO)** | `/data/euc_ero_data_01/` | 600 GB | [ERO Data Product Description](https://www.cosmos.esa.int/web/euclid/ero-public-release) |

### Euclid Astroquery

Use the pre-installed [Euclid Astroquery package](https://astroquery.readthedocs.io/en/latest/esa/euclid/euclid.html) to query the Archive and identify relevant data products. The examples demonstrate cone searches, catalogue queries and product access.

Use the `datalabs_path` values returned by the relevant Archive queries to locate products on the mounted data volumes. Consult each notebook for the query and product-specific workflow.
